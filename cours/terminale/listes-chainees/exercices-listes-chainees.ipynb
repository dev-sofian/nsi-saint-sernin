### Rappel du cours

Une liste chaînée est une suite de **maillons**. Chaque maillon contient une **valeur** et connaît
le maillon **suivant**. La liste elle-même ne retient que sa **tête** (le premier maillon) ;
le dernier maillon a pour suivant `None`.

```
tete
 │
 ▼
┌───┬───┐    ┌───┬───┐    ┌───┬──────┐
│ 8 │ ●─┼──► │ 5 │ ●─┼──► │ 3 │ None │
└───┴───┘    └───┴───┘    └───┴──────┘
```

On reprend **l'implémentation du cours** (classes `Maillon` et `Liste`), avec deux petits outils
pour écrire les tests : `construire` et `en_liste_python`.

### Mode d'emploi

- **Exécute d'abord la cellule ci-dessous**, sinon rien ne marchera.
- Pour chaque exercice : complète la cellule avec `...`, exécute-la, puis exécute la cellule de **tests**.
- Les exercices vont du 🟢 (guidé) au 🔴 (plus difficile). Le corrigé est caché sous chaque exercice.

````python
class Maillon:
    """
    Crée un maillon de la liste chaînée
    """

    def __init__(self, val, suiv):
        self.valeur = val
        self.suivant = suiv


class Liste:
    """
    Crée une liste chaînée
    """

    def __init__(self):
        # self.tete référence un Maillon
        self.tete = None

    def est_vide(self):
        return self.tete is None

    def ajouter(self, val):
        self.tete = Maillon(val, self.tete)


def construire(valeurs):
    """Outil pour les tests : construire([1, 2, 3]) donne la liste chaînée 1 -> 2 -> 3."""
    lst = Liste()
    for v in reversed(valeurs):
        lst.ajouter(v)
    return lst


def en_liste_python(lst):
    """Outil pour les tests : renvoie les valeurs de la liste chaînée dans une liste Python."""
    valeurs = []
    m = lst.tete
    while m is not None:
        valeurs.append(m.valeur)
        m = m.suivant
    return valeurs


print("Classes chargées ✔")
````

## ✅ Bilan

| Opération | Liste chaînée | Tableau (liste Python) |
|---|---|---|
| Ajouter en tête | temps **constant** | temps linéaire (il faut tout décaler) |
| Lire l'élément de rang n | temps **linéaire** (on parcourt n maillons) | temps constant (`t[n]`) |
| Ajouter à la fin | linéaire (sauf si on retient aussi le dernier maillon) | constant en moyenne (`append`) |

- Pour parcourir une liste chaînée : `m = lst.tete`, puis `m = m.suivant` tant que `m is not None`.
- Une liste chaînée est **récursive** par nature : la suite d'un maillon est encore une liste (exercice 7).
- Ajouter et retirer en tête, c'est exactement ce dont on a besoin pour une **pile**.
