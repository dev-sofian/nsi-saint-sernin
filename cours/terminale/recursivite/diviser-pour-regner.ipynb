##  Diviser pour régner : un algorithme récursif

- **cas limite** : le tableau est de taille minimale
- sinon:

   - rechercher le milieu du tableau,
   - appel récursif sur chaque sous-tableau
   - fusionner les tableaux lors de la remontée d’appel

### Étape de la fusion
Lors de la remontée d’appel, la *fusion* assemble deux tableaux triés, en un seul

Pour assembler les deux parties du tableau, il faut prendre le
plus petit élément, jusqu’à vider un des deux blocs. Puis on
complète avec les éléments restants.



__fusionner_

Puis on complète avec les éléments restant  



  
_Compléter_

`````python
from random import randint
from time import time
def tri_fusion(tab:list, deb: int, fin: int)-> None:
    """- rechercher le **milieu** du tableau,
      - **appel récursif** sur chaque **sous-tableau**
      - **fusionner** les tableaux lors de la remontée d’appel"""
    if deb<fin:    
        milieu=(deb+fin)//2
        tri_fusion(tab,deb,milieu)
        tri_fusion(tab,milieu+1,fin)
        fusionner(tab,deb,fin)

        
def fusionner(tab: list, deb: int, fin: int)-> None:
    res=[]
    milieu=(deb+fin)//2
    i=deb 
    j=milieu+1
    #Assembler
    while i <= milieu and j <= fin:
        if tab[i] <= tab[j]:
            res.append(tab[i])
            i += 1
        else:
            res.append(tab[j])
            j += 1
         
    #Compléter
    for i1 in range(i, milieu+1):
        res.append(tab[i1])
    for j1 in range(j, fin+1):
        res.append(tab[j1])
        
     # remplacement tab par res
    for k in range(len(res)):
        tab[deb+k] = res[k]
        
       
tab = [randint(1, 10000) for _ in range(10000)]
deb = time()
tri_fusion(tab, 0, len(tab)-1)
fin = time()
print("fusion ", fin-deb)
`````

<h3> <i class='fa fa-cogs' style="color: darkorange"> Activité 7 </h3>

#### Consigne
À chaque appel de la fonction tri_fusion nous divisons la liste en deux.  
Combien de fois faut-il couper la liste en deux pour obtenir des listes d’un élément 

Mathématiquement nous cherchons $a$ tel que : 
$$\large\frac{n}{2^a}=1 $$

Le logarithme base 2 noté $log_2$ se définit : $log_2(2^x) = x   

$$\large\frac{n}{2^a}=1 $$  

$$\large\Longleftrightarrow n=2^a$$  

$$\large\Longleftrightarrow log_2n=log_2(2^a)$$  

$$\large\Longleftrightarrow  log2 n = a$$

## Principe
### Première approche

Diviser pour régner se décompose en trois parties :
- _diviser_ : Le problème est partagé en plusieurs petits
problèmes identiques.
- _traitement_ : Chaque petit problème est résolu.
- _recombinaison_ : Les petits problèmes résolus sont
assemblés pour remonter au problème principal.
