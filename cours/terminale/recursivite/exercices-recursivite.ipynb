### Rappel du cours

Une fonction **récursive** s'appelle elle-même. Elle a toujours :

1. une **condition d'arrêt** (le cas de base), qui renvoie un résultat sans nouvel appel ;
2. un **appel récursif** sur un problème **plus petit**, qui se rapproche de la condition d'arrêt.

C'est l'exemple du cours :

```python
def somme_recursif(n):
    if n == 0:                          # condition d'arrêt
        return 0
    return n + somme_recursif(n - 1)    # appel récursif sur un problème plus petit
```

## 🟡 Exercice 8 : trouve l'erreur

Ces deux fonctions ne s'arrêtent jamais pour certains appels : Python finit par lever
`RecursionError: maximum recursion depth exceeded`.

```python
def mystere_a(n):
    return n + mystere_a(n - 1)

def mystere_b(n):
    if n == 0:
        return 0
    return n + mystere_b(n - 2)
```

1. Pour chacune, explique **pourquoi** elle ne s'arrête pas.
2. Pour `mystere_b`, donne un appel qui fonctionne et un appel qui plante.
3. Corrige `mystere_b` pour qu'elle s'arrête pour tout entier `n >= 0`.

<details>
<summary><b>Voir la réponse</b></summary>

1. `mystere_a` n'a **aucune condition d'arrêt** : elle s'appelle indéfiniment.
   `mystere_b` a une condition d'arrêt (`n == 0`), mais en enlevant 2 à chaque fois,
   un `n` impair passe par 1, puis −1, −3… et **saute** 0 sans jamais l'atteindre.
2. `mystere_b(4)` fonctionne (4 → 2 → 0) ; `mystere_b(3)` plante (3 → 1 → −1 → …).
3. Une correction possible : arrêter dès que `n` est trop petit.

```python
def mystere_b(n):
    if n <= 0:
        return 0
    return n + mystere_b(n - 2)
```

**À retenir** : il ne suffit pas d'avoir une condition d'arrêt, il faut être sûr que les appels
**finissent par l'atteindre**. C'est la question de la **terminaison**.

</details>

```python
            + hanoi(n - 1, intermediaire, arrivee, depart))     # 3. les n-1 par-dessus
```

```python
def dichotomie(liste, valeur, debut, fin):
    if debut > fin:                          # zone de recherche vide : pas trouvé
        return False
    milieu = (debut + fin) // 2
    if liste[milieu] == valeur:
        return True
    if valeur < liste[milieu]:               # on cherche dans la moitié gauche
        return dichotomie(liste, valeur, debut, milieu - 1)
    return dichotomie(liste, valeur, milieu + 1, fin)   # sinon dans la moitié droite
```

## ✅ Bilan

- Une fonction récursive a une **condition d'arrêt** et un **appel récursif sur un problème plus petit**.
- Il faut vérifier que les appels **atteignent forcément** la condition d'arrêt (exercice 8), sinon : `RecursionError`.
- Chaque appel en attente occupe une place sur la **pile d'appels** (exercice 1).
- Une récursion qui recalcule les mêmes choses peut être très lente : la **mémoïsation** y remédie (exercice 10).
- Couper un problème en deux à chaque appel, c'est **diviser pour régner** (exercice 12).
