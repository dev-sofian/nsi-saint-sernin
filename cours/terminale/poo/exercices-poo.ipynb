### Exercice 2 — Comparer deux approches *(évaluer)*

Voici deux implémentations d'une même classe `Temperature`. Analyse-les et rédige une comparaison argumentée.

````python
# 🔴 Exercice 2 — Version A : accès direct

class TemperatureA:
    def __init__(self, valeur):
        self.valeur = valeur  # en Celsius

t_a = TemperatureA(-300)   # aucune vérification
print(t_a.valeur)
````

````python
# 🔴 Exercice 2 — Version B : encapsulée avec @property

class TemperatureB:
    def __init__(self, valeur):
        self._valeur = valeur

    @property
    def valeur(self):
        return self._valeur

    @valeur.setter
    def valeur(self, v):
        if v < -273.15:
            raise ValueError("Température impossible (en-dessous du zéro absolu)")
        self._valeur = v

try:
    t_b = TemperatureB(-300)
except ValueError as e:
    print(f"Erreur : {e}")
````

## Exercice 9 
**Question** : quel est l’ordre d’appel des constructeurs ? 

````python
class A:
    def __init__(self):
        print("Init A")

class B(A):
    def __init__(self):
        super().__init__()
        print("Init B")

class C(A):
    def __init__(self):
        super().__init__()
        print("Init C")

class D(B, C):
    def __init__(self):
        super().__init__()
        print("Init D")

d = D()
````
