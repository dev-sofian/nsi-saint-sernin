## 📚 Explication 3 — Lire des syntaxes avancées

### 🎯 Objectif de ce bloc
> Identifier une syntaxe Python inconnue (lambda, map/filter, générateurs), en chercher le sens, et l'expliquer avec ses propres mots.

Certaines syntaxes sont plus rares mais se rencontrent dans du code réel. Face à une syntaxe totalement inconnue, la méthode est :  
-  1) isoler le fragment, 
- 2) le tester seul dans l'interpréteur, 
- 3) reformuler avec ses propres mots.


#### lambda 

Le mot-clé lambda en Python fournit un raccourci pour déclarer de petites fonctions anonymes. Les fonctions lambda se comportent comme des fonctions normales déclarées avec le mot-clé def. Elles peuvent être utilisées chaque fois que des objets de type fonction sont requis.  
Par exemple, voici comment vous définiriez une fonction lambda simple effectuant une addition :

```python
addition = lambda x, y: x + y
 addition(5, 3)
#8
```
Vous pouvez déclarer la même fonction addition avec le mot-clé def :
```python
def addition(x, y):
    return x + y
addition(5, 3)
#8
```

```python
nombres = [1, 2, 3, 4]
doubles = list(map(lambda x: x * 2, nombres))   # applique lambda à chaque élément
print(doubles)                     # [2, 4, 6, 8]
```



#### Générateur  

Un **générateur** ressemble à une fonction, mais utilise `yield` au lieu de `return` : il produit ses valeurs **une par une**, à la demande, plutôt que de toutes les calculer d'un coup.
```python
def compteur(n):
    for i in range(n):
        yield i

for valeur in compteur(3):
    print(valeur)   # affiche 0, puis 1, puis 2
```

**Point de vigilance** : `map`, `filter` et les générateurs ne renvoient pas directement une liste — il faut souvent les envelopper dans `list(...)` pour voir leur contenu d'un coup, ou les parcourir avec une boucle `for`.

Vérification : pourquoi `print(compteur(3))` n'affiche-t-il pas `0 1 2`, mais quelque chose comme `<generator object ...>` ?
